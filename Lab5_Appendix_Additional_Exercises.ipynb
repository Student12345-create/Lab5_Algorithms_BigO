```python
# STUDENT SPECIFIC INPUT RULES
LAST_NAME = "LORILLA"
STUDENT_ID = "TUPM-26-1923"
FAVORITE_ARTIST = "SKUSTA CLEE"

# LAST DIGIT OF THE STUDENT ID
digits = [int(char) for char in STUDENT_ID if char.isdigit()]

SEED_DIGIT = digits[-1]
ID_SUM = sum(digits)

# LENGTH OF THE LAST NAME
NAME_LENGTH = len(LAST_NAME.strip())

# LENGTH OF THE FAVORITE ARTIST NAME
ARTIST_LENGTH = len(FAVORITE_ARTIST.replace(" ", ""))

# ALPHABET VALUE OF EACH LETTER
ARTIST_SUM = 0

for letter in FAVORITE_ARTIST.upper():
    if letter.isalpha():
        ARTIST_SUM += ord(letter) - ord('A') + 1

print("STUDENT SPECIFIC INPUT RULES")
print("Seed Digit:" , SEED_DIGIT)
print("ID Sum:" , ID_SUM)
print("Name Length:" , NAME_LENGTH)
print("Artist Length:" , ARTIST_LENGTH)
print("Artist Sum:" , ARTIST_SUM)
```

```python
# EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH
# Incident record
original_data = []

for i in range(12):
    incident = 1000 + (
        (ID_SUM * 3 + ARTIST_SUM + NAME_LENGTH + i * 17) % 9000
    )
    original_data.append(incident)

# ROTATION
rotation = SEED_DIGIT % len(original_data)

rotated_data = original_data[rotation:] + original_data[:rotation]

# Generate the target using Id sum and Artist sum
target_index = (ID_SUM + ARTIST_SUM) % len(rotated_data)
target = original_data[target_index]

print("EXERCISE 1: GENERATED DATA")
print("Original Dataset:", original_data)
print("Rotation Point:", rotation)
print("Rotated Dataset:", rotated_data)
print("Target Incident:", target)
```

```python
# BINARY SEARCH
def rotated_binary_search(data, target):
    low = 0
    high = len(data) - 1
    comparisons = 0

    while low <= high: 
        mid = (low + high) // 2
        comparisons += 1

        if data[mid] == target:
            return mid, comparisons

        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1

        else:
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1

    return -1, comparisons

binary_index, binary_comparisons = rotated_binary_search(rotated_data, target)

print(" BINARY SEARCH")
print("Target:", target)

if binary_index != -1:
    print("Index:", binary_index)
    print("Position:", binary_index + 1)
else:
    print("Target incident was not found.")
print("Comparisons/Probes:", binary_comparisons)
```

```python
# LINEAR SEARCH
def linear_search(data, target):
    comparisons = 0

    for index, value in enumerate(data):
        comparisons += 1
        if value == target:
            return index, comparisons

    return -1, comparisons

linear_search_index, linear_search_comparisons = linear_search(rotated_data, target)
print(" LINEAR SEARCH")
print("Target:", target)

if linear_search_index != -1:
    print("Index:", linear_search_index)
    print("Position:", linear_search_index + 1)
else:
    print("Target incident was not found.")
print("Comparisons/Probes:", linear_search_comparisons)
print("\nSEARCH COMPARISON")
print("Binary Search Comparisons:", binary_index)
print("Linear Search:", linear_search_comparisons)
print("Both found the target:", binary_index == linear_search_index)
```

```python
# EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS
#GENERATE STUDENT RECORDS FOR DEPARTMENT A
department_a = []

for i in range(5):
    student_id = 10000 + SEED_DIGIT * 100 +  ID_SUM + i * 3
    department_a.append(student_id)

# GENERATE STUDENT RECORDS FOR DEPARTMENT B
department_b = []

for i in range(5):
    student_id = ( 30000 + ARTIST_LENGTH * 100 + NAME_LENGTH + 1 * 5)
    department_b.append(student_id)

print("Department A:", department_a)
print("Department B:", department_b)

def merge_sorted_lists(list_a, list_b):
    merged = []
    i = 0
    j = 0
    comparisons = 0

    #compare

    while i < len(list_a) and j < len(list_b):
        comparisons += 1
        if list_a[i] <= list_b[j]:
            merged.append(list_a[i])
            i += 1
        else:
            merged.append(list_b[j])
            j += 1
        return merged, comparisons

merged_data, merge_comparisons = merge_sorted_lists(department_a, department_b)

print("\nMERGED ENROLLMENT RECORDS")
print("Consolidated Dataset:", merged_data)
before_count = len(department_a) + len(department_b)
after_count = len(merged_data)

print("Records Before:", before_count)
print("Records After:", after_count)
print("Merge Comparisons:", merge_comparisons)
print("Ascending Order Verified:", merged_data == sorted(merged_data))
print("Record Cpount Verified:", before_count == after_count)
```

```python
# EXERCISE 3: E-Commerce Order Processing
import random 
import time
# ORDERS
orders = []

for i in range(8):
    order_number = 100 + ((ID_SUM + ARTIST_SUM + NAME_LENGTH + i * 17) % 900)
    orders.append(order_number)

sorted_input = sorted(orders)
reverse_input = sorted(orders, reverse=True)

random_input = orders.copy()
random_generator = random.Random(ID_SUM + ARTIST_SUM)
random_generator.shuffle(random_input)

print("GENERATED ORDER DATASET:", orders)
print("Sorted Input:", sorted_input)
print("Reverse Input:", reverse_input)
print("Random Input:", random_input)
```

```python
# BUBBLE SORT
def bubble_sort(data):
    a = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(a) - 1):
        swapped = False

        for j in range(len(a) - 1 - i):
            comparisons += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                movements += 1
                swapped = True

            if not swapped:
                break

    return a, comparisons, movements
```

```python
# SELECTION SORT
def selection_sort(data):
    a = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(a) - 1):
        minimum = 1

        for j in range(i + 1, len(a)):
            comparisons += 1

            if a[j] < a[minimum]:
                minimum = j
            
            if minimum != i:
                a[i], a[minimum] = a[minimum], a[i]
                movements += 1
        
        return a, comparisons, movements
```

```python
# INSERTION SORT
def insertion_sort(data):
    a = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if a [j] <= key:
                break 
            a[j + 1] = a[j]
            movements += 1
            j -= 1

        a[j + 1] = key

    return a, comparisons, movements
    
```

```python
# COMPARING ALL ALGORITHMS
def measure_sort(sort_function, data, trials=5):
    elapsed_time = []
    result = None

    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        elapsed = time.perf_counter() - start
        elapsed_time.append(elapsed)
    average_time = sum(elapsed_time) / trials

    return result, average_time

algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

input_conditions = {
    "Sorted": sorted_input,
    "Reverse": reverse_input,
    "Random": random_input 
}

print("SORTING PERFORMANCE RESULTS")
print("-" * 95)
print(
    f"{'Input':<10}{'Algorithm':<18}"
    f"{'Comparisons':>14}{'Movements':>12}"
    f"{'Time (seconds)':>20}{'Correct':>12}"
)
print("-" * 95)

for input_name, data in input_conditions.items():
    for algorithm_name, sort_function in algorithms.items():
        result, average_time = measure_sort(sort_function, data)

        sorted_result = result[0]
        comparisons = result[1]
        movements = result[2]

        correct = sorted_result == sorted(data)

        print(
            f"{input_name:<10}{algorithm_name:<18}"
            f"{comparisons:>14}{movements:>12}"
            f"{average_time:>20.8f}{str(correct):>12}"
        )

print("=" * 95)




```
